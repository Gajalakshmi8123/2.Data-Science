test the analysis of variance between etest_p and mba_p at signifiance level 5%

```python
from scipy.stats import ttest_rel
male=Dataset[Dataset['gender']=='M']['etest_p']
male1=Dataset[Dataset['gender']=='M']['mba_p']
ttest_rel(male,male1)
```


This is the t-statistic — a standardized value that compares the difference between group means relative to variability.

A large t-statistic (positive or negative) indicates a strong difference between the groups or between a sample mean and a known mean.

The p-value tells you how likely this result is under the null hypothesis (no difference).

There is a very strong statistical difference between the group means.

The null hypothesis (no difference) is rejected with extremely high confidence.

This result is unlikely due to random chance.

test the similaruty between the degree_t(sci & tech) and specialization(mk&HR) with respect to salary at signifance level at 5%

```python
from scipy.stats import ttest_ind
Dataset=Dataset.dropna()
male=Dataset[Dataset['degree_t']=='Sci&Tech']['salary']
female=Dataset[Dataset['specialisation']=='Mkt&HR']['salary']
ttest_ind(male,female)
```

There is a statistically significant difference between the two groups (p ≈ 0.008). The observed difference is unlikely due to chance."

convert the normal distribution into standard normal distribution for salary column

```python
def StdNBgraph(Dataset):
    import seaborn as sns
    mean=Dataset.mean()
    std=Dataset.std()
    values=[i for i in Dataset]
    z_score=[((j-mean)/std) for j in values]
    sns.distplot(z_score,kde=True)
    sum(z_score)/len(z_score)
    
```
